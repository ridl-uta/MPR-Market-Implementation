```python
import pandas as pd
import matplotlib.pyplot as plt

# ACM-ish styling
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Linux Libertine O', 'Libertine', 'Times New Roman', 'Times', 'DejaVu Serif'],
    'mathtext.fontset': 'stix',
    'font.size': 8,
    'axes.labelsize': 8,
    'axes.titlesize': 8,
    'axes.linewidth': 0.8,
    'lines.linewidth': 1.0,
    'lines.markersize': 3,
    'legend.fontsize': 7,
})
df = pd.read_csv('Data/sample_execution_times.csv')

# Plot 1: execution time vs clients (no title; caption in paper)
fig, ax = plt.subplots(figsize=(3.33, 2.0), dpi=150)
ax.plot(df['execution_time_sec'], df['clients'], marker='o', color='black')
ax.set_xlabel('Execution Time (s)')
ax.set_ylabel('Number of Clients')
ax.grid(True, which='major', linestyle='--', linewidth=0.4, alpha=0.4)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
fig.tight_layout()
plt.show()

# Plot 2: communication/computation time vs clients (no title)
fig, ax = plt.subplots(figsize=(3.33, 2.0), dpi=150)
ax.plot(df['communication_time_sec'], df['clients'], marker='o', color='black', linestyle='-')
ax.plot(df['computation_time_sec'], df['clients'], marker='s', color='gray', linestyle='--')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Number of Clients')
ax.legend(['Communication', 'Computation'], frameon=False)
ax.grid(True, which='major', linestyle='--', linewidth=0.4, alpha=0.4)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
fig.tight_layout()
plt.show()
```

```python
# Export figures to files
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'Times', 'DejaVu Serif'],
    'font.size': 10,
    'axes.labelsize': 10,
    'axes.titlesize': 11,
    'axes.linewidth': 0.8,
    'lines.linewidth': 1.2,
    'lines.markersize': 4,
    'legend.fontsize': 9,
})

df = pd.read_csv('Data/sample_execution_times.csv')

# Figure 1
fig, ax = plt.subplots(figsize=(3.5, 2.6), dpi=300)
ax.plot(df['execution_time_sec'], df['clients'], marker='o', color='black')
ax.set_xlabel('Execution Time (s)')
ax.set_ylabel('Number of Clients')
ax.set_title('Clients vs. Execution Time')
ax.grid(True, which='major', linestyle='--', linewidth=0.5, alpha=0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
fig.tight_layout()
fig.savefig('Data/clients_vs_execution_time.png', dpi=300)
plt.close(fig)

# Figure 2
fig, ax = plt.subplots(figsize=(3.5, 2.6), dpi=300)
ax.plot(df['communication_time_sec'], df['clients'], marker='o', color='black', linestyle='-')
ax.plot(df['computation_time_sec'], df['clients'], marker='s', color='gray', linestyle='--')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Number of Clients')
ax.set_title('Clients vs. Communication/Computation Time')
ax.legend(['Communication', 'Computation'], frameon=False)
ax.grid(True, which='major', linestyle='--', linewidth=0.5, alpha=0.5)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
fig.tight_layout()
fig.savefig('Data/clients_vs_comm_comp_time.png', dpi=300)
plt.close(fig)

print('Saved: Data/clients_vs_execution_time.png')
print('Saved: Data/clients_vs_comm_comp_time.png')
```
